The helper the note above refers to:


````python
def root_ids(series):
    """Root ids as a plain uint64 array, safe to hand to a CAVE query."""
    return series.dropna().astype("uint64").to_numpy()
````

**Coordinates come in three flavours and mixing them fails silently.** Voxels, raw
nanometres, and *transformed* microns (rotated so the pia is flat, with y as cortical depth).
The last is the most meaningful for geometry — and subtracting a raw-µm position from a
transformed-µm one adds a large fixed offset with no error and no warning. Push every
transform through one helper so it cannot happen:

````python
def to_transformed_um(df, prefix):
    """Columns f'{prefix}_x/y/z' in nanometres -> transformed microns, pia-flat."""
    cols = [f"{prefix}_{ax}" for ax in "xyz"]
    return v1dd_ds.transform("nm").apply(df[cols].values)
````

**A synapse's position relative to the cell it lands on.** Question 3 will ask you to compute the location of a synapse relative to the cell body in spherical coordinates. Feel free to write your own function, but rather than get hung up on geometry you can also use this function to accomplish that. Given a set of points and a soma, both in transformed µm, it returns how far each point is from that soma and in what
direction — with the compass flattened to a single angle.


````python
def r_theta(points_um, soma_um):
    """Spherical-ish coordinates about a soma, with azimuth dropped.

    r     : straight-line distance from the soma, µm
    theta : degrees out of the tangential plane; +90 = straight toward the pia,
            0 = level with the soma, -90 = straight down.

    Dropping azimuth assumes the target is roughly rotationally symmetric about the
    pia axis -- fair for a pyramidal cell, wrong for anything with a planar arbor.
    """
    delta = points_um - soma_um
    lateral = np.hypot(delta[:, 0], delta[:, 2])
    return (
        np.linalg.norm(delta, axis=1),
        np.degrees(np.arctan2(-delta[:, 1], lateral)),
    )
````

**Plotting that frame.** `cloud` draws one panel of synapses in a target's own frame. It is
here so the figures later stay short; read it when you first use it in Question 4.


````python
def cloud(ax, df, title, color):
    """Every synapse in the frame of the cell it landed on. Azimuth was collapsed, so
    lateral distance is a magnitude -- mirror it, which is what the rotational-symmetry
    assumption already claims."""
    lat = df.r_um * np.cos(np.radians(df.theta_deg))
    vert = df.r_um * np.sin(np.radians(df.theta_deg))
    ax.scatter(
        np.r_[lat, -lat], np.r_[vert, vert], s=3.0, alpha=0.18, lw=0, color=color
    )
    ax.scatter([0], [0], marker="*", s=180, c="k", zorder=6)
    for rad in (25, 50, 100, 150):
        ax.add_artist(plt.Circle((0, 0), rad, fill=False, color="0.6", lw=0.7, ls=":"))
    ax.set(
        xlim=(-160, 160),
        ylim=(-160, 200),
        aspect="equal",
        title=title,
        xlabel="lateral distance (µm)",
    )
    # The frame is clipped so the tight cloud is readable; say how much falls outside
    # rather than letting the crop quietly imply the cloud ends there.
    out = 100 * (df.r_um > 160).mean()
    if out > 1:
        ax.text(
            0.97,
            0.03,
            f"{out:.0f}% beyond frame",
            transform=ax.transAxes,
            ha="right",
            va="bottom",
            fontsize=8,
            color="0.4",
        )
````

## Question 1 — Who are the L5-ET cells, and whose reconstruction do you have to trust?

**ET** is *extratelencephalic*: They are a kind of pyramidal cell that projects out of cortex, to thalamus,
superior colliculus and brainstem. They are also known as thick-tufted cells, with a long apical trunk
rising to a (sometimes very wide) tuft in Layer 1, plus a basal skirt around the soma.

Inhibitory cells come pre-sorted into four subclasses — `PTC`, `DTC`, `ITC`, `STC` — by
`cell_type_multifeature_v1`, a classifier built on **soma, nucleus, dendrite and spine
morphology**. Note what is *not* in that list: the classifier never sees where a cell's
synapses land.

**Treat the four labels as opaque for now.** They are four groups that look different down a
microscope, nothing more. Whether they also *behave* differently — whether they put their
synapses in different places on a target — is the question this module aims to answer. The
names do stand for something, and we come back to them at the end; unpacking them now would
hand you the answer before you have measured anything.


**Task:** Build a table of all classified cells from `cell_type_multifeature_v1` and remove any duplicate root ids that do not have a well-defined type. How many of each type, and which are excitatory vs inhibitory?

Note: The table query function is `client.materialize.query_table`, or you can also use `client.materialize.tables.<table>.query`

*Solution note: Later cells expect this table to be called `cell_df`.*

````python
cell_df = client.materialize.tables.cell_type_multifeature_v1().get_all(
    split_positions=True, desired_resolution=[1, 1, 1]
)
# Cells appearing twice are ambiguous merges; drop both copies rather than picking one.
cell_df = cell_df.drop_duplicates(subset="pt_root_id", keep=False)

print(f"{len(cell_df):,} classified cells")
print(cell_df.classification_system.value_counts().to_string(), "\n")
print(cell_df.cell_type.value_counts().to_string())
````

Note the spelling: **`L5ET`, no hyphen.** Other modules in this workshop load derived data
products that spell the same type `L5-ET`.

The shared palette in `utils` is keyed the hyphenated way too, so bridge the two spellings
once rather than maintaining a second dict.

````python
import re  # Regex is a tool to search for and replace strings based on matching various patterns.


def hyphenated(t):
    """CAVE spells excitatory types 'L5ET'; cell_type_palette keys them 'L5-ET'."""
    return re.sub(r"^(L\d)([A-Z]+)$", r"\1-\2", str(t))


PALETTE = {
    t: cell_type_palette.get(hyphenated(t), "0.6")
    for t in cell_df.cell_type.dropna().unique()
}
print({k: PALETTE[k] for k in list(PALETTE)[:4]})
````

Soma depth by cell type is worth plotting before going further — it is the sanity check that the coordinate transform is doing what it claims.

````python
xyz = to_transformed_um(cell_df, "pt_position")
cell_df["depth_um"] = xyz[:, 1]
cell_df["tx_um"], cell_df["tz_um"] = xyz[:, 0], xyz[:, 2]

order = [
    "L2IT",
    "L3IT",
    "L4IT",
    "L5IT",
    "L5ET",
    "L5NP",
    "L6IT",
    "L6CT",
    "L6SP",
    "PTC",
    "DTC",
    "ITC",
    "STC",
]
order = [t for t in order if t in set(cell_df.cell_type.dropna())]

fig, ax = plt.subplots(figsize=(9, 4.6))
sns.violinplot(
    data=cell_df[cell_df.cell_type.isin(order)],
    x="cell_type",
    y="depth_um",
    hue="cell_type",
    order=order,
    hue_order=order,
    palette=PALETTE,
    cut=0,
    linewidth=0.6,
    inner=None,
    legend=False,
    ax=ax,
)
ax.invert_yaxis()
ax.set(
    xlabel="",
    ylabel="soma depth below pia (µm)",
    title="Figure 1: Cortical layering falls out of soma depth",
)
for x in (4.5, 8.5):
    ax.axvline(x, color="0.7", lw=0.8, ls="--")
ax.text(2.0, 40, "excitatory", ha="center", color="0.35", fontsize=9)
ax.text(10.5, 40, "inhibitory", ha="center", color="0.35", fontsize=9)
sns.despine()
plt.tight_layout()
plt.show()
````

The excitatory types stack in layer order — that is the sanity check that the coordinate
transform is doing what it claims. The inhibitory types do not, because they are defined by

### Whose reconstruction has to be right?

Now the part that governs every number in this module.

**Task:** Build the presynaptic universe. Join `proofreading_status_and_strategy` to the cell
type table and count, per inhibitory subclass, how many cells have `status_axon == True`.
Then look at `strategy_axon` and ask what the partially-extended cells are.

*Call the joined, filtered frame `inh_pf`.*

````python
pf_df = client.materialize.tables.proofreading_status_and_strategy().get_all(
    split_positions=True
)

pf_ct = pf_df.merge(
    cell_df[["pt_root_id", "classification_system", "cell_type"]],
    on="pt_root_id",
    how="inner",
)

inh_pf = pf_ct.query("status_axon == True and classification_system == 'inhibitory'")

print(f"{len(pf_df):,} cells in the proofreading table")
print(f"{(pf_df.status_axon == True).sum():,} with a proofread axon")
print(f"{len(inh_pf):,} of those are inhibitory\n")
print(pd.crosstab(inh_pf.cell_type, inh_pf.strategy_axon).to_string())
````

**Questions — no code. This is the conceptual core of the module.**

You are about to study synapses *onto* an L5-ET cell.

1. **Whose reconstruction quality matters, the ET cell's or the interneurons'?**
   Both, differently. The ET cell needs a trustworthy **dendrite**, which is what makes "we
   found every synapse onto it" true. The interneurons need trustworthy **axons**, which is
   what makes "this synapse came from a PTC" true.

2. **A trusted dendrite means every synapse was found. Does it mean every synapse was
   correctly attributed?**
   No. Segmentation makes **merge errors** — a fragment of another cell's axon glued onto a
   soma. The synapse is real and its location is right; the presynaptic identity is wrong.
   Proofreading removes those merges, which is why `status_axon == True` is the filter used
   throughout. Although note that small numbers of errors will remain after proofreading, so
   always double check that any result isn't based on some small number of potential artifacts.

4. **Reconstruction loses thin distal branches more often than thick proximal ones. Suppose
   one subclass tends to synapse close to its targets' somata and another travels far out
   along the dendrite. Without the filter, which of the two looks over-represented?**
   The proximal one. Truncation strips the far-travelling subclass's synapses while leaving
   the close one's largely intact, so an unfiltered analysis reports "inhibition is proximal"
   **whatever the biology does.** That is the specific reason this module filters — and note
   you cannot yet say which label is which, which is the point. Question 2 measures the size
   of the effect; Question 3 identifies the players.

5. **What does the restriction cost?**
   About 95% of the typed inhibitory cells (Figure 2). A much smaller sample, in exchange for
   labels that mean what they say.

`strategy_axon` further splits proofread cells into fully and partially extended, but that is
**completeness, not correctness** — a partially extended axon has had its merges cleaned, so
its synapses are still correctly attributed. Completeness only matters when you need a
per-axon denominator, which no question here does.

````python
inh_proofread_ids = root_ids(inh_pf.pt_root_id)
inh_type = inh_pf.set_index("pt_root_id").cell_type  # lookup used from here on
print(f"{len(inh_proofread_ids)} proofread inhibitory axons carried forward")
````

## Question 2 — What lands where?

Synapses are labelled by what they land on: a dendritic **spine**, a dendritic **shaft**, or
the **soma**. The distinction is electrotonic. A spine is a small, electrically isolated
compartment, so excitatory input there can summate locally. A synapse on the soma sits right
where the axon initial segment decides whether to fire, so it can shunt the whole cell.
The textbook claims, stated so we can check them:

1. Excitatory → excitatory synapses are onto **spines**.
2. Inhibitory → excitatory synapses are onto **shafts**.
3. Synapses onto inhibitory cells are onto **shafts**.
4. Basket cells target **somata** — so one of the four inhibitory labels should own most
   somatic synapses. Which one?


**Task — assemble the synapse table.** Three joins get you from "a list of L5-ET cells" to
"one row per synapse, with a compartment and a presynaptic subclass": pull the input
synapses, attach the compartment label, attach the presynaptic cell type and restrict to
proofread axons. Report the synapse count before and after that last step.

*The next three cells do them in order — read each one before running it, because the third
is a decision and the first two are only plumbing.*

*Names later cells rely on: `COHORT` and `syn` for the 15-cell query, `COHORT_WIDE`
and `syn_inh` for the two-sided one.*

````python
l5et = cell_df.query("cell_type == 'L5ET'").copy()

# Take a set of 15 cells NEAREST the exemplar, not an arbitrary slice. An interneuron's axon
# spans a few hundred microns, and proofreading effort is itself spatially clustered,
# so a cohort spread across the volume gets ~5 proofread inhibitory synapses per cell
# and leaves some with none at all. A local one gets a couple of hundred.
ex = l5et.loc[l5et.pt_root_id == EXEMPLAR_L5ET].iloc[0]
l5et["lateral_um"] = np.hypot(l5et.tx_um - ex.tx_um, l5et.tz_um - ex.tz_um)
l5et = l5et.sort_values("lateral_um")

COHORT = root_ids(l5et.pt_root_id[:15])
print(
    f"{len(l5et):,} L5ET cells; cohort of {len(COHORT)} within "
    f"{l5et.lateral_um.iloc[:15].max():.0f} um of the exemplar"
)

syn = client.materialize.synapse_query(
    post_ids=COHORT, desired_resolution=[1, 1, 1], split_positions=True
)
print(f"{len(syn):,} input synapses onto {len(COHORT)} L5ET cells")
````

The compartment label lives in `synapse_target_predictions_ssa`, a reference table keyed on
synapse id. Note it is `_ssa`, not `_ssa_v2` — only the former exists at materialization 1196.

`fetch_tags` splits the request into chunks, which is given here rather than explained —
there is a reason the server will not take it all at once, and Question 3 makes you find it.


````python
def fetch_tags(syn_ids, chunk=40_000):
    """Compartment label (spine / shaft / soma) for each synapse id."""
    out = []
    for i in range(0, len(syn_ids), chunk):
        out.append(
            client.materialize.query_table(
                "synapse_target_predictions_ssa",
                filter_in_dict={"target_id": list(map(int, syn_ids[i : i + chunk]))},
                select_columns=["target_id", "tag"],
                merge_reference=False,
                log_warning=False,
            )
        )
    return pd.concat(out, ignore_index=True).rename(columns={"target_id": "id"})
````

````python
tags = fetch_tags(syn.id.astype("int64").to_numpy())
syn = syn.merge(tags, on="id", how="left")
print(
    f"labelled {syn.tag.notna().sum():,} of {len(syn):,} "
    f"({100 * syn.tag.notna().mean():.1f}%)"
)
print(syn.tag.value_counts().to_string())
````

Now the presynaptic side, and **the one real choice in this sequence**: restricting to cells
with a proofread axon. This restricts the analysis set quite a lot, but still leaves a lot of measured inputs.

````python
syn["pre_type"] = syn.pre_pt_root_id.map(cell_df.set_index("pt_root_id").cell_type)
syn["pre_class"] = syn.pre_pt_root_id.map(
    cell_df.set_index("pt_root_id").classification_system
)

n_all = len(syn)
n_typed = syn.pre_type.notna().sum()
syn_pf = syn[
    syn.pre_pt_root_id.isin(inh_proofread_ids) | (syn.pre_class == "excitatory")
].copy()

print(f"all input synapses            {n_all:>8,}")
print(f"  ... from a typed cell       {n_typed:>8,}  ({100 * n_typed / n_all:.0f}%)")
print(
    f"  ... after proofread filter  {len(syn_pf):>8,}  ({100 * len(syn_pf) / n_all:.0f}%)"
)
````

Most input synapses have **no presynaptic label at all** — orphan axon fragments with no soma
in the volume, so there is nothing to type them by. They are real synapses we simply cannot
attribute, and what follows is conditional on the presynaptic cell being identifiable, which is
*not* a random subset of all input.

The set below narrows the query at *both* ends — `pre_ids` and `post_ids` together are a
logical AND — so we can sweep far more L5-ET cells while paying only for the synapses we want.
**It is for the inhibitory subclasses alone.** It comes from a different cohort with a
different filter than the excitatory synapses above, so counts from the two are not on the
same footing. Mixing them is the trap two cells down.

````python
COHORT_WIDE = root_ids(l5et.pt_root_id[:250])  # still nearest-first

syn_inh = client.materialize.synapse_query(
    pre_ids=inh_proofread_ids,
    post_ids=COHORT_WIDE,
    desired_resolution=[1, 1, 1],
    split_positions=True,
)
syn_inh["tag"] = syn_inh.id.map(
    fetch_tags(syn_inh.id.astype("int64").to_numpy()).set_index("id").tag
)
syn_inh["pre_type"] = syn_inh.pre_pt_root_id.map(inh_type)

print(
    f"{len(syn_inh):,} proofread-inhibitory synapses onto {len(COHORT_WIDE)} L5ET cells"
)
print(syn_inh.pre_type.value_counts().to_string())
````

**Task:** Build the contingency table of presynaptic type × target compartment, and
normalize it **by row**. This is a table where each row will be a cell type, each column will be a compartment, and the value will be the number of synapses in that compartment. You can then ask why the column version — "who owns this compartment?" — is
not a question you can answer here.

````python
inh_types = ["PTC", "DTC", "ITC", "STC"]

# One row per presynaptic group. Rows are normalized independently, so the excitatory
# row (15-cell cohort, typed) and the inhibitory rows (250-cell cohort, proofread) can
# sit in the same table even though they come from different queries -- see below.
exc = syn[(syn.pre_class == "excitatory") & syn.tag.notna()]
inh = syn_inh[syn_inh.pre_type.isin(inh_types) & syn_inh.tag.notna()]

counts = pd.concat(
    [
        pd.crosstab(pd.Series("excitatory (all)", index=exc.index), exc.tag),
        pd.crosstab(inh.pre_type, inh.tag),
    ]
).reindex(index=["excitatory (all)"] + inh_types, columns=["spine", "shaft", "soma"])
counts.insert(0, "n", counts.sum(axis=1))
counts
````

````python
by_row = counts[["spine", "shaft", "soma"]].div(counts.n, axis=0) * 100

fig, ax = plt.subplots(figsize=(5.4, 3.4))
sns.heatmap(
    by_row,
    annot=True,
    fmt=".1f",
    cmap="Blues",
    cbar=False,
    vmin=0,
    vmax=100,
    ax=ax,
    annot_kws={"size": 9},
)
ax.set(
    xlabel="",
    ylabel="",
    title="Figure 3: where each presynaptic type puts its synapses (% of row)",
)
plt.tight_layout()
plt.show()
````

**Read this table across the rows!**

A row asks **where a given type puts its synapses.** It is conditional on the type, so it does
not care how well any *other* type happens to be sampled — which is why the excitatory row and
the inhibitory rows can come from different queries and still both be sound. Each row is its
own denominator.

Note that because we are limiting our analysis to cell typed neurons with proofread axons, so this
leaves out a lot of cells, not only including non-proofread cells, but also pyramidal cells or other
projections from outside the volume such as thalamic axons. This means that this tables does not answer
the complementary question of where a given compartment *recieves* synapses from due to this imbalance.

Against the textbook, from the rows alone:

1. **Excitatory → spine holds** — around 79% of attributable excitatory input. Published
   figures run higher; the proximal dendrite and soma of an L5-ET are genuinely sparsely
   spined, and the classifier misses some spine necks.
2. **Inhibitory → shaft holds**, and increasingly so along `PTC` → `DTC` → `ITC` → `STC`.
3. Untestable here — our postsynaptic cells are all excitatory by construction.
4. **Only `PTC` targets somata to any real extent** — around 16% of its synapses, against
   roughly 1% for the other three. That is claim 4, made safely from a row: not "somata are
   mostly PTC" (a column claim we cannot support) but "PTC is the one subclass that puts a
   substantial share of its output on somata."

Worth noticing in passing: the somatic share falls away sharply after `PTC` while the shaft
share climbs steadily to `STC`. That ordering shows up again in Question 3, measured
geometrically instead of from labels.

Before the statistics, one picture. Here are the exemplar's input synapses on its own
morphology, coloured by presynaptic subclass — one cell, not the result.

## Question 3 — Where along the dendrite?

The right measure is **geodesic distance**: path length through the dendrite from the soma. It

## Question 4 — Build a cheap frame, and find out what it costs

Question 3 did it the right way, and the cost is visible: one skeleton, one cell. The next
question needs *hundreds* of target cells, so geodesic distance is out of reach.

The way round is a frame anchored on the target's **soma position**, which is just a column —
`r_theta` from the setup. But it is an approximation, so before spending it we measure what it
costs, on the one cell where both measures exist.

**Task 1 — how good is straight-line distance?** Compute **r** for the same synapses and
scatter it against the geodesic distance you already have.

**Predict first, in writing:** as synapses get further from the soma, does the approximation
get better or worse?

````python
root_xyz = dend.s.root_location
soma_um = v1dd_ds.transform("nm").apply(root_xyz.reshape(1, 3))[0]
syn_um = v1dd_ds.transform("nm").apply(anno.vertices)
syn_df["r_um"], syn_df["theta_deg"] = r_theta(syn_um, soma_um)

# Does theta mean what we think? There is no apical label in this dataset to check
# against, so colour the skeleton by it and look.
vert_um = v1dd_ds.transform("nm").apply(dend.skeleton.vertices)
_, vert_theta = r_theta(vert_um, soma_um)

fig, ax = plt.subplots(figsize=(5.2, 5.5))
osy.plot.plot_morphology_2d(
    dend.transform(v1dd_ds.transform("nm").apply),
    projection="xy",
    color="compartment",
    palette={1: "0.3", 3: "0.8"},
    linewidth=0.7,
    ax=ax,
)
ax.scatter(
    vert_um[:, 0],
    vert_um[:, 1],
    s=1.4,
    c=np.where(vert_theta > 30, "#B4472E", "#0E6E78"),
    lw=0,
)
ax.set(
    title="Figure 5: θ > 30° (red) picks out the apical, with no apical label in the data",
    xlabel="µm",
    ylabel="depth (µm)",
)
sns.despine()
plt.tight_layout()
plt.show()
````

````python
d = syn_df[syn_df.r_um > 1].copy()
d["tortuosity"] = d.geodesic_um / d.r_um

r_pearson = np.corrcoef(d.geodesic_um, d.r_um)[0, 1]
r_spear = stats.spearmanr(d.geodesic_um, d.r_um).statistic
print(f"Pearson r = {r_pearson:.4f}   Spearman = {r_spear:.4f}")
print(
    f"tortuosity: median {d.tortuosity.median():.3f}, "
    f"p10 {d.tortuosity.quantile(0.1):.3f}, p90 {d.tortuosity.quantile(0.9):.3f}"
)
````

````python
fig, axes = plt.subplots(1, 2, figsize=(10.5, 4))
axes[0].scatter(d.r_um, d.geodesic_um, s=1.5, alpha=0.12, c="#0E6E78", lw=0)
lim = [0, max(d.r_um.max(), d.geodesic_um.max()) * 1.02]
axes[0].plot(lim, lim, "k--", lw=1, label="perfect agreement")
axes[0].set(
    xlim=lim,
    ylim=lim,
    xlabel="straight-line r (µm)",
    ylabel="geodesic distance (µm)",
    aspect="equal",
    title=f"Pearson r = {r_pearson:.3f}",
)
axes[0].legend(frameon=False, fontsize=9)

band_edges = [0, 50, 100, 200, 400, 1000]
labels = [f"{a}–{b}" for a, b in zip(band_edges[:-1], band_edges[1:])]
d["band"] = pd.cut(d.r_um, band_edges, labels=labels)
med = d.groupby("band", observed=True).tortuosity.median()
axes[1].bar(range(len(med)), med.values, color="#0E6E78", width=0.65)
axes[1].axhline(1, color="k", lw=1, ls="--")
axes[1].set_xticks(range(len(med)))
axes[1].set_xticklabels(med.index, fontsize=9)
axes[1].set(
    xlabel="straight-line distance band (µm)",
    ylabel="median tortuosity",
    ylim=(0.95, max(1.5, med.max() * 1.12)),
    title="does the error grow with distance?",
)
for i, v in enumerate(med.values):
    axes[1].text(i, v + 0.012, f"{v:.2f}", ha="center", fontsize=9)
fig.suptitle(
    "Figure 6: the approximation is worst where you least expected",
    y=1.02,
    x=0.02,
    ha="left",
    fontsize=11,
)
sns.despine()
plt.tight_layout()
plt.show()
````

**At least here, error shrinks with distance.**

The reason is geometry. A synapse 400 µm out is on the apical tuft, reached by a nearly
straight trunk — path length is close to the straight line. A synapse 30 µm out is on a
twisting local basal branch that wanders half again as far as the direct route. So the
approximation is *best exactly where the compartment story is most interesting*, and worst
close in, within a few tens of microns of the soma.

That last clause is a genuine caveat, not a reassurance: proximal r is systematically an
underestimate of electrotonic distance. It does not threaten a *comparison* between subclasses,
which is what Question 4 makes, but it would threaten a quantitative claim about decay.


**Task 2 — does the ranking survive?** Recompute the subclass comparison with **r** and

**Choosing the exemplars.** Filter on `status_axon` for correctness as always, then

samples fewer of its targets and possibly a skewed subset.


````python
soma_lookup = cell_df.set_index("pt_root_id")[
    ["tx_um", "depth_um", "tz_um", "cell_type", "classification_system"]
]

fully = pf_ct.query(
    "strategy_axon == 'axon_fully_extended' and classification_system == 'inhibitory'"
)
print(fully.cell_type.value_counts().to_string())
````

**Task 3 — start with one of each.** Pull the output synapses of one `PTC` and one `DTC`, and
place every synapse in the frame of *the target it landed on*. Plot the two side by side.

One cloud on its own is uninterpretable — a shape only means something against a comparison.
Two cells cost two queries and about a second, and they give you the answer before any of the
plumbing that follows.

````python
EXEMPLAR_PTC = int(root_ids(fully.query("cell_type == 'PTC'").pt_root_id)[0])
EXEMPLAR_DTC = int(root_ids(fully.query("cell_type == 'DTC'").pt_root_id)[0])


def outputs_in_target_frame(pre_ids):
    """Every synapse these cells make, in (r, theta) about each target's own soma."""
    s = client.materialize.synapse_query(
        pre_ids=np.atleast_1d(pre_ids).astype("uint64"),
        desired_resolution=[1, 1, 1],
        split_positions=True,
    )
    s = s.join(soma_lookup, on="post_pt_root_id", how="inner")
    if len(s) == 0:
        return s
    pts = v1dd_ds.transform("nm").apply(
        s[["ctr_pt_position_x", "ctr_pt_position_y", "ctr_pt_position_z"]].values
    )
    # r_theta takes one soma; here every synapse has its own, so subtract first and
    # hand it a zero origin.
    s["r_um"], s["theta_deg"] = r_theta(
        pts - s[["tx_um", "depth_um", "tz_um"]].values, np.zeros(3)
    )
    return s
````

````python
one = {
    t: outputs_in_target_frame(rid)
    for t, rid in [("PTC", EXEMPLAR_PTC), ("DTC", EXEMPLAR_DTC)]
}
for t, d in one.items():
    print(
        f"{t}: {len(d):,} synapses onto {d.post_pt_root_id.nunique():,} typed targets, "
        f"median r = {d.r_um.median():.0f} um, theta = {d.theta_deg.median():+.0f} deg"
    )


fig, axes = plt.subplots(1, 2, figsize=(11, 4.6), sharey=True)
for ax, t in zip(axes, ["PTC", "DTC"]):
    cloud(
        ax,
        one[t],
        f"one {t} — median r = {one[t].r_um.median():.0f} µm",
        PALETTE.get(t, "k"),
    )
axes[0].set_ylabel("← below soma      toward pia → (µm)")
fig.suptitle(
    "Figure 8: two interneurons, each synapse in its target's own frame",
    y=1.0,
    x=0.02,
    ha="left",
    fontsize=11,
)
sns.despine()
plt.tight_layout()
plt.show()
````

**There it is, on two cells.** One label sits tight against the soma; the other sits several
times further out and leans toward the pia. Neither cloud knows anything about *which* target
each synapse landed on — that was divided out by construction.

Two cells is not a result, though. The rest of this question asks whether they were
representative — which needs twenty, and twenty is where the server starts saying no.


**Task 4 — now scale it the obvious way, and watch it break.** You need this for ~20
interneurons. Put all twenty ids into a single query. **Write down what you expect before
you run it.**

````python
many = root_ids(fully.query("cell_type == 'PTC'").pt_root_id)[:40]
try:
    client.materialize.synapse_query(
        pre_ids=many, desired_resolution=[1, 1, 1], split_positions=True
    )
    print("...that worked, which it usually does not")
except Exception as e:
    print(f"{type(e).__name__}: {str(e)[:160]}")
````

**Two things to take from that failure.**

**It is not a bug in your code and not a bad root id.** The traceback ends in
`requests.HTTPError`, not a `KeyError` — your request was well-formed and the *server* refused
it. A 502 with HTML in the body is a proxy giving up on a response that got too large.
Recognising which layer failed is the transferable skill.

**The limit is on response size, not id count.** Twenty of these cells return fine; forty do
not. Twenty cells with bigger axons might also not. So "find a batch size that works" is a
fragile fix — pick one comfortably inside the limit and handle failure gracefully.

**Task 5 — fix it.** Batch the request and concatenate. Ten ids per call sits inside the
limit and costs a tenth of the round trips; since `synapse_query` returns `pre_pt_root_id` on
every row, a `groupby` recovers per-cell results afterwards. The design choice worth making
deliberately is what to do when a batch *does* fail — discarding all ten is wasteful when
usually nine are fine.


````python
import time


def collect(pre_ids, chunk=10, pause=3.0):
    """Query in batches, falling back to one-at-a-time when a batch fails.

    Ten ids per request sits inside the limit and costs a tenth of the round trips.
    If a batch does fail, retrying it cell-by-cell isolates the bad one instead of
    discarding all ten. Returns (frame, ids that never came back).
    """
    frames, failed = [], []
    for i in range(0, len(pre_ids), chunk):
        batch = [int(p) for p in pre_ids[i : i + chunk]]
        try:
            frames.append(outputs_in_target_frame(batch))
        except Exception:
            time.sleep(pause)
            for pid in batch:
                try:
                    frames.append(outputs_in_target_frame([pid]))
                except Exception:
                    failed.append(pid)
    out = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()
    return out, failed


N_PER_SUBCLASS = 8
panel = {}
for t in ["PTC", "DTC"]:
    ids = root_ids(fully.query("cell_type == @t").pt_root_id)[:N_PER_SUBCLASS]
    df, failed = collect(ids)
    df["pre_type"] = t
    panel[t] = df
    print(
        f"{t}: {len(df):,} synapses from {df.pre_pt_root_id.nunique()} cells"
        + (f"  [failed: {failed}]" if failed else "")
    )

allpre = pd.concat(panel.values(), ignore_index=True)
````

Two refinements in there worth insisting on, because they are what separates a script that
survives a real dataset from one that does not. **Retry with a pause**, because a 502 can also
be transient and a twenty-minute loop that dies on iteration 17 has wasted twenty minutes. And
**collect failures rather than raising**, so one bad cell costs you one cell rather than the
whole run.

**Task 7 — the repeatability question.** Pooling over targets could still hide the thing we
actually want to know. Split by target: for each (interneuron, target) pair compute a median r,
and look at the spread *within* one interneuron. Does a single cell do the same thing to all of
its targets, or is it proximal on some and distal on others?

Make a dataframe where each row is a proofread interneuron (with known type) and there are columns for the median radial distance to soma as well as the interquartile interval (25-to-75th percentile range) of the radial distance distribution.
Plot 1) a histogram of the median distance distribution by interneuron subclass and 2) a plot showing median and IQR values for each interneuron as a dot with errorbars.

````python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8))
rows = []
for t in ["PTC", "DTC"]:
    df = panel[t]
    per_t = (
        df.groupby(["pre_pt_root_id", "post_pt_root_id"])
        .r_um.agg(["size", "median"])
        .query("size >= 5")
    )
    axes[0].hist(
        per_t["median"],
        bins=np.arange(0, 260, 10),
        histtype="step",
        lw=2,
        density=True,
        color=PALETTE.get(t, "k"),
        label=f"{t}: {len(per_t):,} target cells",
    )
    for pre, g in per_t.groupby("pre_pt_root_id"):
        rows.append(
            {
                "pre": pre,
                "subclass": t,
                "n_targets": len(g),
                "median_r": g["median"].median(),
                "iqr": g["median"].quantile(0.75) - g["median"].quantile(0.25),
            }
        )
axes[0].set(
    xlabel="median r for one target cell (µm)",
    ylabel="density",
    title="each observation is one (interneuron, target) pair",
)
axes[0].legend(frameon=False, fontsize=9)

per_cell = pd.DataFrame(rows)
for i, (t, g) in enumerate(per_cell.groupby("subclass")):
    axes[1].errorbar(
        g.median_r,
        np.arange(len(g)) + i * 10,
        xerr=g.iqr / 2,
        fmt="o",
        color=PALETTE.get(t, "k"),
        ms=5,
        lw=1.2,
        capsize=2,
        label=t,
    )
axes[1].set(
    xlabel="median r across that cell's targets (µm)",
    yticks=[],
    title="one row per interneuron, bar = IQR across its targets",
)
axes[1].legend(frameon=False, fontsize=9)
fig.suptitle(
    "Figure 9: targeting is a property of the presynaptic cell",
    y=1.02,
    x=0.02,
    ha="left",
    fontsize=11,
)
sns.despine()
plt.tight_layout()
plt.show()
per_cell.round(1)
````

**This is the answer to the question the module is named after.** A single PTC touches
hundreds of different target cells and does substantially the same thing to each: its
per-target median r has an IQR of a few tens of microns, against a PTC-vs-DTC separation
several times larger. The subclass label is not a summary of a heterogeneous bag — it is
predicting a real, repeatable property of individual cells.

**Task 8 — split by target cell type.** Does an interneuron treat every kind of target

### Optional — the denominator problem

*Not needed for anything above, but useful for thinking about the data.*

Every distance histogram in this module is partly a picture of **where the dendrite is**. More
synapses at 100 µm may just mean more dendrite at 100 µm. It did not distort the comparisons
here, because two subclasses measured on the same cells share the same denominator and it
largely cancels — but the moment you compare *across* cells, or ask an absolute question like
"how dense is inhibition on the tuft", it stops cancelling.

The fix on a cell with a skeleton is to divide by **µm of dendrite per bin**. Run the cell
below and compare the middle panel to the right one: the raw counts are dominated by the shape
of the denominator.

````python
bins = np.arange(0, 601, 25)
inh = syn_df[syn_df.is_pf_inh & syn_df.pre_type.isin(inh_types)]
DASH = {"PTC": "-", "DTC": "-", "ITC": "--", "STC": ":"}

vert_dist = dend.skeleton.distance_to_root() / 1_000.0
vert_len = dend.skeleton.half_edge_length / 1_000.0
dendrite_um, _ = np.histogram(vert_dist, bins=bins, weights=vert_len)
centers = 0.5 * (bins[:-1] + bins[1:])

fig, axes = plt.subplots(1, 3, figsize=(13, 3.6))
axes[0].bar(centers, dendrite_um, width=22, color="0.8")
axes[0].set(
    xlabel="geodesic distance (µm)", ylabel="µm of dendrite", title="the denominator"
)
for t in inh_types:
    g = inh[inh.pre_type == t].geodesic_um
    if len(g) < 5:
        continue
    n, _ = np.histogram(g, bins=bins)
    axes[1].step(
        centers, n, where="mid", lw=2, color=PALETTE.get(t, "k"), ls=DASH[t], label=t
    )
    with np.errstate(divide="ignore", invalid="ignore"):
        dens = np.where(dendrite_um > 20, n / dendrite_um, np.nan)
    axes[2].step(
        centers, dens, where="mid", lw=2, color=PALETTE.get(t, "k"), ls=DASH[t], label=t
    )
axes[1].set(xlabel="geodesic distance (µm)", ylabel="synapses", title="raw count")
axes[2].set_yscale("log")
axes[2].set(
    xlabel="geodesic distance (µm)",
    ylabel="synapses per µm of dendrite",
    title="density — the same data, divided (log)",
)
axes[2].legend(frameon=False, fontsize=9)
fig.suptitle(
    "Optional figure: every count needs a denominator",
    y=1.02,
    x=0.02,
    ha="left",
    fontsize=11,
)
sns.despine()
plt.tight_layout()
plt.show()
````

**Where this goes next.** Two threads worth a project:

1. The denominator here is *this cell's* dendrite. To compare across cells you need a
   denominator that travels — excitatory synapse density is one candidate, since spiny
   dendrite carries them at a roughly steady rate, and it needs no skeleton at all.
2. Everything above treats "distance from soma" as the axis. Electrotonically what matters is
   path length *and* diameter. `cell.skeleton` carries radius; a cable-weighted distance would
   be a genuinely better axis and nobody has run it on this dataset.

### Optional — is the signature a property of the cell, the subclass, or the pair?

*Harder than it looks, and the reason is a lesson about summary statistics.*


````python
def ratio(df, label):
    within = df.groupby("subclass").median_r.std().mean()
    between = df.groupby("subclass").median_r.mean().std()
    print(
        f"{label:26s} within {within:5.1f} um | between {between:5.1f} um "
        f"| ratio {between / within:4.2f}"
    )


ratio(per_cell, "all cells")
ratio(per_cell.query("n_targets >= 30"), "cells with >=30 targets")

print()
print(
    per_cell.sort_values(["subclass", "median_r"])[
        ["subclass", "pre", "n_targets", "median_r", "iqr"]
    ].to_string(index=False)
)
````

**Read the ratio, then read the table underneath it, and notice they disagree.**

Taken at face value the ratio is unimpressive — barely above 1, which would say subclass
predicts almost nothing. But scan the per-cell table: most PTCs sit tightly in the teens and
twenties of microns, the DTCs between roughly 60 and 90. Those groups barely overlap.

**Two cells are doing the damage, and they are the two worst-sampled ones** — a PTC out at
DTC-like distances and a DTC in at PTC-like distances, both estimated from **eighteen
targets** against one to four hundred for everyone else. Because `std` squares deviations,
two bad estimates sink the ratio by themselves. Requiring thirty targets moves it from about
**1.2 to nearly 9.**

That is not p-hacking; it is declining to average in estimates you already know are
unreliable. Decide the threshold before you look, and say you applied one.

**The habit worth taking away is to plot before you summarise.** Figure 9 showed the
separation immediately; no variance ratio was needed to see it. Then ask the biological
question: is that odd cell mis-typed, badly reconstructed, or a real variant?

## Summary

1. **Excitatory input arrives on spines across the whole dendrite; inhibitory input arrives on
   shaft and soma** — and the soma is essentially inhibitory-only real estate.
2. **Which inhibitory subclass a synapse comes from predicts where it lands.** PTC within a
   few tens of microns of the soma, DTC several times further out and leaning apical.
3. **That is a property of the presynaptic cell.** One PTC does the same thing to hundreds of
   different targets.

And three habits worth more than the result: **filter for the failure mode your question
actually has**; **give every count a denominator** (Figure 11); **calibrate your shortcuts**
(Figure 7 is why Figure 8 is allowed to exist).

### What the names meant

Now they can be unpacked. `PTC` is **perisomatic targeting**, `DTC` **dendrite targeting**,
`ITC` **inhibitory targeting**, `STC` **sparse targeting** — the nomenclature of
Schneider-Mizell et al. The classifier that assigned them never saw a synapse; it worked from
soma, nucleus, dendrite and spine morphology alone. So what you measured is that morphology
predicts targeting well enough that the names are earned.

Two things that buys you. First, it is a real prediction rather than a definition — the labels
could have failed to separate, and you would have seen that. Second, it puts a number on
Module 1's caveat that these labels "might align more with molecular class than exact
targeting for a given neuron, particularly among ITC/VIP cells." Your `ITC` count — nine
synapses on the exemplar — is exactly where that caveat bites.

### Where students get stuck, in order of how often

1. **`L5ET` vs `L5-ET`.** Deliberate. Other modules load derived products that hyphenate,
   so a copied filter returns an empty frame with no error. Let them hit it — the fix is
   one character and the lesson is "check your filter returned something."
2. **`pt_root_id` is a nullable `Int64`.** Passing `.values` into a query raises
   `TypeError: Object of type IntegerArray is not JSON serializable`. `root_ids()` is given
   for exactly this reason; students who write their own casting hit it.
3. **Mixed coordinate frames in Q3 Task 1.** Soma positions and synapse positions must both
   go through `v1dd_ds.transform("nm")` before subtracting. Dividing nanometres by 1000
   gives raw microns, and subtracting the two frames adds a silent fixed offset of a few
   hundred µm. **The tell is that every subclass comes out identical at ~375 µm.** If a
   student reports "PTC and DTC look the same", check this first — it is the one bug in the
   module that produces a confident-looking null result.
4. **Forgetting the dendrite mask.** The exemplar's axon is about a third of its skeleton.
   Leaving it in corrupts geodesic distances and the length denominator.
5. **`status_axon` vs `strategy_axon`.** Expect confusion. `status_axon` gates correctness
   (merges cleaned); `strategy_axon` describes completeness. Everything here needs the
   former; only per-axon denominators need the latter.
