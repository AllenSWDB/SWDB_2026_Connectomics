**Coordinates come in three flavours and mixing them fails silently.** Voxels, raw
nanometres, and *transformed* microns (rotated so the pia is flat, with y as cortical depth).
The last is the most meaningful for geometry — and subtracting a raw-µm position from a
transformed-µm one adds a large fixed offset with no error and no warning. Push every
transform through one helper so it cannot happen:

````python
def to_transformed_um(df, prefix):
    """Columns f'{prefix}_x/y/z' in nanometres -> transformed microns, pia-flat."""
    cols = [f"{prefix}_{ax}" for ax in "xyz"]
    return v1dd_ds.transform("nm").apply(df[cols].values)
````

**A synapse's position relative to the cell it lands on.** Question 3 will ask you to compute the location of a synapse relative to the cell body in spherical coordinates. Feel free to write your own function, but rather than get hung up on geometry you can also use this function to accomplish that. Given a set of points and a soma, both in transformed µm, it returns how far each point is from that soma and in what
direction — with the compass flattened to a single angle.


````python
def r_theta(points_um, soma_um):
    """Spherical-ish coordinates about a soma, with azimuth dropped.

    r     : straight-line distance from the soma, µm
    theta : degrees out of the tangential plane; +90 = straight toward the pia,
            0 = level with the soma, -90 = straight down.

    Dropping azimuth assumes the target is roughly rotationally symmetric about the
    pia axis -- fair for a pyramidal cell, wrong for anything with a planar arbor.
    """
    delta = points_um - soma_um
    lateral = np.hypot(delta[:, 0], delta[:, 2])
    return (
        np.linalg.norm(delta, axis=1),
        np.degrees(np.arctan2(-delta[:, 1], lateral)),
    )
````

**Plotting that frame.** `cloud` draws one panel of synapses in a target's own frame. It is
here so the figures later stay short; read it when you first use it in Question 4.


````python
def cloud(ax, df, title, color):
    """Every synapse in the frame of the cell it landed on. Azimuth was collapsed, so
    lateral distance is a magnitude -- mirror it, which is what the rotational-symmetry
    assumption already claims."""
    lat = df.r_um * np.cos(np.radians(df.theta_deg))
    vert = df.r_um * np.sin(np.radians(df.theta_deg))
    ax.scatter(
        np.r_[lat, -lat], np.r_[vert, vert], s=3.0, alpha=0.18, lw=0, color=color
    )
    ax.scatter([0], [0], marker="*", s=180, c="k", zorder=6)
    for rad in (25, 50, 100, 150):
        ax.add_artist(plt.Circle((0, 0), rad, fill=False, color="0.6", lw=0.7, ls=":"))
    ax.set(
        xlim=(-160, 160),
        ylim=(-160, 200),
        aspect="equal",
        title=title,
        xlabel="lateral distance (µm)",
    )
    # The frame is clipped so the tight cloud is readable; say how much falls outside
    # rather than letting the crop quietly imply the cloud ends there.
    out = 100 * (df.r_um > 160).mean()
    if out > 1:
        ax.text(
            0.97,
            0.03,
            f"{out:.0f}% beyond frame",
            transform=ax.transAxes,
            ha="right",
            va="bottom",
            fontsize=8,
            color="0.4",
        )
````

**A few concepts you might have started to think about**

You are about to study synapses *onto* an L5-ET cell.

1. **Whose reconstruction quality matters, the ET cell's or the interneurons'?**
   Both, differently. The ET cell needs a trustworthy **dendrite**, which is what makes "we
   found every synapse onto it" true. The interneurons need trustworthy **axons**, which is
   what makes "this synapse came from a PTC" true.

2. **A trusted dendrite means every synapse was found. Does it mean every synapse was
   correctly attributed?**
   No. Segmentation makes **merge errors** — a fragment of another cell's axon glued onto a
   soma. The synapse is real and its location is right; the presynaptic identity is wrong.
   Proofreading removes those merges, which is why `status_axon == True` is the filter used
   throughout. Although note that small numbers of errors will remain after proofreading, so
   always double check that any result isn't based on some small number of potential artifacts.

4. **Reconstruction loses thin distal branches more often than thick proximal ones. Suppose
   one subclass tends to synapse close to its targets' somata and another travels far out
   along the dendrite. Without the filter, which of the two looks over-represented?**
   The proximal one. Truncation strips the far-travelling subclass's synapses while leaving
   the close one's largely intact, so an unfiltered analysis reports "inhibition is proximal"
   **whatever the biology does.** That is the specific reason this module filters — and note
   you cannot yet say which label is which, which is the point. Question 2 measures the size
   of the effect; Question 3 identifies the players.

5. **What does the restriction cost?**
   About 95% of the typed inhibitory cells (Figure 2). A much smaller sample, in exchange for
   labels that mean what they say.

## Question 2 — What lands where?

Synapses are labelled by what they land on: a dendritic **spine**, a dendritic **shaft**, or
the **soma**. The distinction is electrotonic. A spine is a small, electrically isolated
compartment, so excitatory input there can summate locally. A synapse on the soma sits right
where the axon initial segment decides whether to fire, so it can shunt the whole cell.
The textbook claims, stated so we can check them:

1. Excitatory → excitatory synapses are onto **spines**.
2. Inhibitory → excitatory synapses are onto **shafts**.
3. Synapses onto inhibitory cells are onto **shafts**.
4. Basket cells target **somata** — so one of the four inhibitory labels should own most
   somatic synapses. Which one?

### What the names meant

Now they can be unpacked. `PTC` is **perisomatic targeting**, `DTC` **dendrite targeting**,
`ITC` **inhibitory targeting**, `STC` **sparse targeting** — the nomenclature of
Schneider-Mizell et al. The classifier that assigned them never saw a synapse; it worked from
soma, nucleus, dendrite and spine morphology alone. So what you measured is that morphology
predicts targeting well enough that the names are earned.

Two things that buys you. First, it is a real prediction rather than a definition — the labels
could have failed to separate, and you would have seen that. Second, it puts a number on
Module 1's caveat that these labels "might align more with molecular class than exact
targeting for a given neuron, particularly among ITC/VIP cells." Your `ITC` count — nine
synapses on the exemplar — is exactly where that caveat bites.
